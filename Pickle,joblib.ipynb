```python
from sklearn.linear_model import LinearRegression
import pickle
import pandas as pd
from joblib import dump,load
data = pd.read_csv('/content/homeprices.csv')
df=pd.DataFrame(data)
df
x=df[['area']]
y=df['price']
model = LinearRegression()
model.fit(x,y)
dump(model,"homeprice_model.joblib")
with open("homeprice_modelpk1","wb") as f:
  pickle.dump(model,f)
m1=load("homeprice_model.joblib")
print(m1.predict(pd.DataFrame([[3000]],columns=['area'])))
with open("homeprice_modelpk1","rb") as f:
  m2=pickle.load(f)
print(m2.predict(pd.DataFrame([[3000]],columns=['area'])))
```
