```python
import pandas as pd
from sklearn.linear_model import LinearRegression
import matplotlib.pyplot as plt
data=pd.read_csv("/content/carprices.csv")
df=pd.DataFrame(data)

plt.scatter(df['Mileage'],df['Sell Price($)'],color="blue",marker='+')
plt.title("CAR SELLING PRICESS")
plt.xlabel('Mileage')
plt.ylabel('Sell Price($)')
plt.savefig('carrrrrsssssssssss')
plt.show()


plt.scatter(df['Age(yrs)'],df['Sell Price($)'],color="red",marker='x')
plt.title("years pf buying carssssss")
plt.xlabel('Age(yrs)')
plt.ylabel('Sell Price($)')
plt.savefig('agessssss')
plt.clf()

dum=pd.get_dummies(df['Car Model']).astype(int)
mer=pd.concat([df,dum],axis='columns')
f=mer.drop(['Car Model','Mercedez Benz C class'],axis='columns')
print(f.head())

X=f.drop('Sell Price($)',axis='columns')
Y=f['Sell Price($)']
print(list(X.columns))
model = LinearRegression()
model.fit(X,Y)
acc=model.score(X,Y)
print(f"model score:{acc:.4f}(or {acc*100:.2f}%)")
m1=model.predict(pd.DataFrame([[4500,4,0,0]],columns = X.columns))
print(f"predict of benz: ${m1[0]:,.2f}")
m2=model.predict(pd.DataFrame([[12300,6,0,0]],columns = X.columns))
print(f"predict of bmw : ${m2[0]:,.2f}")
```
